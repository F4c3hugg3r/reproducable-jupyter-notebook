# Pipeline Step 1: Data Cleaning

**Goal:** This module loads the raw sensor data, filters out physically impossible values (spikes), and saves the cleaned dataset for later visualization.

**Reproducibility Note:** We use fixed random seeds for data generation and pass the threshold dynamically via `papermill`.

```python
# --- CONFIGURATION / PARAMETERS ---
# This cell must receive the 'parameters' tag so Papermill can overwrite the values.
THRESHOLD = 50.0  
RANDOM_SEED = 42
OUTPUT_FILE = "data/cleaned_data.csv"
```

```python
import pandas as pd
import numpy as np

# 1. Guarantee reproducibility of randomness
np.random.seed(RANDOM_SEED)

# 2. Simulate raw data (In practice: pd.read_csv("raw_data.csv"))
days = pd.date_range(start="2026-01-01", periods=100)
raw_values = np.random.normal(loc=20.0, scale=15.0, size=100)
# Add some extreme error values (spikes)
raw_values[10] = 95.0 
raw_values[45] = 120.0

df_raw = pd.DataFrame({'date': days, 'sensor_value': raw_values})

# 3. Filter based on our parameter
df_cleaned = df_raw[df_raw['sensor_value'] <= THRESHOLD].copy()

print(f"Original data points: {len(df_raw)}")
print(f"Data points after filtering (Threshold <= {THRESHOLD}): {len(df_cleaned)}")

# 4. Save intermediate result for the next notebook
df_cleaned.to_csv(OUTPUT_FILE, index=False)
```
