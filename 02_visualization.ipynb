# Pipeline Step 2: Visualization

**Goal:** This notebook reads the dataset generated by `01_data_cleaning.ipynb` and creates the final plot. By strictly separating cleaning and visualization (modularization), the process remains organized and reproducible.

```python
import pandas as pd
import matplotlib.pyplot as plt

INPUT_FILE = "data/cleaned_data.csv"

# Load the cleaned data
try:
    df = pd.read_csv(INPUT_FILE, parse_dates=['date'])
except FileNotFoundError:
    print(f"Error: File '{INPUT_FILE}' not found. Did you run Notebook 01 first?")
    raise

# Create the plot
plt.figure(figsize=(10, 5))
plt.plot(df['date'], df['sensor_value'], marker='o', linestyle='-', color='#1f77b4')
plt.title('Cleaned Sensor Data Over Time')
plt.xlabel('Date')
plt.ylabel('Sensor Value')
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Display the plot in the notebook
plt.show()
```
